```python
import csv
import requests
import time
import hashlib
import os
from tqdm import tqdm  # Added tqdm import
import pandas as pd
# Replace with your VirusTotal API key
API_KEY = 'bb8c073346e504fd1ec60c7997597a2a201ef778a6609cbf6e49f5fd229f7df3'
VT_API_URL = 'https://www.virustotal.com/api/v3/files'
VT_REPORT_URL = 'https://www.virustotal.com/api/v3/analyses/'
```

```python


def calculate_sha256(file_path):
    """Calculate SHA256 hash of a file"""
    try:
        with open(file_path, 'rb') as f:
            sha256 = hashlib.sha256()
            while chunk := f.read(8192):
                sha256.update(chunk)
            return sha256.hexdigest()
    except Exception as e:
        print(f"Error calculating hash for {file_path}: {e}")
        return None

def upload_file(file_path, api_key):
    """Upload file to VirusTotal"""
    headers = {
        'x-apikey': api_key
    }
    
    try:
        with open(file_path, 'rb') as f:
            files = {'file': (os.path.basename(file_path), f)}
            response = requests.post(VT_API_URL, headers=headers, files=files)
            response.raise_for_status()
            return response.json()['data']['id']
    except Exception as e:
        print(f"Error uploading {file_path}: {e}")
        return None

def get_analysis_report(analysis_id, api_key):
    """Get analysis report from VirusTotal"""
    headers = {
        'x-apikey': api_key
    }
    
    try:
        url = f"{VT_REPORT_URL}{analysis_id}"
        while True:
            response = requests.get(url, headers=headers)
            response.raise_for_status()
            result = response.json()
            
            if result['data']['attributes']['status'] == 'completed':
                return result
            time.sleep(15)  # Wait before polling again
    except Exception as e:
        print(f"Error getting report for {analysis_id}: {e}")
        return None

def generate_label(report):
    """Generate label based on VirusTotal results"""
    if not report or 'data' not in report:
        return "ERROR"
    
    stats = report['data']['attributes']['stats']
    malicious = stats.get('malicious', 0)
    suspicious = stats.get('suspicious', 0)
    
    # Define your labeling criteria
    if malicious > 3:  # More than 3 AVs flag as malicious
        return "MALICIOUS"
    elif malicious > 0 or suspicious > 2:  # Any malicious or multiple suspicious
        return "SUSPICIOUS"
    else:
        return "CLEAN"

def process_csv(csv_path):
    """Process CSV file and analyze PE files"""
    # Read CSV with pandas, assuming tab-separated
    df = pd.read_csv(csv_path)
    
    # Ensure column names are correct
    if 'File_Name' not in df.columns or 'Full_Path' not in df.columns:
        raise ValueError("CSV must contain 'File_Name' and 'Full_Path' columns")
    
    results = []
    
    # Iterate through rows with tqdm
    for _, row in tqdm(df.iterrows(), total=len(df), desc="Processing files"):
        file_name = str(row['File_Name']).strip()
        full_path = str(row['Full_Path']).strip()
        
        if not os.path.exists(full_path):
            print(f"File not found: {full_path}")
            results.append((file_name, full_path, "FILE_NOT_FOUND"))
            continue
            
        print(f"Processing: {file_name} ({full_path})")
        
        # Calculate hash
        sha256 = calculate_sha256(full_path)
        if not sha256:
            results.append((file_name, full_path, "HASH_ERROR"))
            continue
            
        # Upload file
        analysis_id = upload_file(full_path, API_KEY)
        if not analysis_id:
            results.append((file_name, full_path, "UPLOAD_ERROR"))
            continue
            
        # Get report
        report = get_analysis_report(analysis_id, API_KEY)
        print(report)
        label = generate_label(report)
        
        results.append((file_name, full_path, label))
        print(f"Result for {file_name}: {label}")
        
        # Respect VirusTotal's rate limit (4 requests/min for free tier)
        time.sleep(15)
    
    # Create results DataFrame and save to CSV
    results_df = pd.DataFrame(results, columns=['File_Name', 'Full_Path', 'Label'])
    results_df.to_csv('vt_results.csv', index=False)
```

```python
# Replace with your CSV file path
csv_file_path = 'APT.csv'
process_csv(csv_file_path)
```
