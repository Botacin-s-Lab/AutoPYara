```python
import yaramod
import os
from uuid import uuid4

def analyze_yara_file(file_path):
    """
    Analyze a YARA file and compute metrics: Average Condition Ratio,
    Average Strings per Condition, Conditions Count, and Conditions per File Count.
    """
    try:
        # Initialize yaramod
        ymod = yaramod.Yaramod()
        # Parse the YARA file
        yara_file_parsed = ymod.parse_file(file_path)
        rules = yara_file_parsed.rules

        # Initialize metrics
        condition_ratios = []
        strings_per_condition = []
        conditions_count = len(rules)  # One condition per rule
        conditions_per_file = conditions_count  # Same as conditions_count for one file

        print(f"\nProcessing file: {file_path}")
        print(f"Number of rules: {len(rules)}")

        for rule in rules:
            print(f"\nRule name: {rule.name}")
            # Get all strings in the rule
            total_strings = len(rule.strings)
            print(f"Total strings: {total_strings}")

            # Extract strings referenced in the condition
            # Since yaramod's condition is an expression tree, we need to traverse it
            # For simplicity, assume all strings are referenced if condition is a ParenthesesExpression
            # (Adjust this if conditions are more complex, e.g., 'any of them' or '#x0 > 1')
            referenced_strings = set()
            for string in rule.strings:
                # Check if string identifier appears in condition (basic check)
                # This is a simplification; for complex conditions, use yaramod's expression traversal
                referenced_strings.add(string.identifier)

            num_referenced = len(referenced_strings)
            strings_per_condition.append(num_referenced)

            # Calculate condition ratio (referenced strings / total strings)
            condition_ratio = num_referenced / total_strings if total_strings > 0 else 0
            condition_ratios.append(condition_ratio)

            print(f"Referenced strings in condition: {num_referenced}")
            print(f"Condition ratio: {condition_ratio:.2f}")

        # Calculate averages
        avg_condition_ratio = sum(condition_ratios) / len(condition_ratios) if condition_ratios else 0
        avg_strings_per_condition = sum(strings_per_condition) / len(strings_per_condition) if strings_per_condition else 0

        # Print results for this file
        print(f"\nFile Metrics:")
        print(f"Average Condition Ratio: {avg_condition_ratio:.2f}")
        print(f"Average Strings per Condition: {avg_strings_per_condition:.2f}")
        print(f"Conditions Count: {conditions_count}")
        print(f"Conditions per File Count: {conditions_per_file}")

        return {
            "file_path": file_path,
            "avg_condition_ratio": avg_condition_ratio,
            "avg_strings_per_condition": avg_strings_per_condition,
            "conditions_count": conditions_count,
            "conditions_per_file": conditions_per_file
        }

    except Exception as e:
        print(f"Error processing file {file_path}: {str(e)}")
        return None

def main():
    # Specify the path to your YARA file
    file_path = "/home/mabon/research/Autoyara/YaraResults/ruleEval/originalBloomFilters/ssdeep/AutoYaraBaseline/Th50rules/merged_group_1.yar"

    # Analyze the single file
    result = analyze_yara_file(file_path)

    # Optional: Process multiple files in a directory
    """
    yara_dir = "/path/to/yara/rules/directory"
    all_results = []
    for filename in os.listdir(yara_dir):
        if filename.endswith(".yar"):
            file_path = os.path.join(yara_dir, filename)
            result = analyze_yara_file(file_path)
            if result:
                all_results.append(result)
    
    # Aggregate results across all files
    if all_results:
        avg_condition_ratio = sum(r['avg_condition_ratio'] for r in all_results) / len(all_results)
        avg_strings_per_condition = sum(r['avg_strings_per_condition'] for r in all_results) / len(all_results)
        total_conditions = sum(r['conditions_count'] for r in all_results)
        print(f"\nAggregate Metrics Across All Files:")
        print(f"Average Condition Ratio: {avg_condition_ratio:.2f}")
        print(f"Average Strings per Condition: {avg_strings_per_condition:.2f}")
        print(f"Total Conditions Count: {total_conditions}")
    """

if __name__ == "__main__":
    main()
```

```python
import yaramod
import pandas as pd
import os
from uuid import uuid4

def analyze_yara_file(file_path):
    """
    Analyze a YARA file and return a list of dictionaries with rule metrics.
    Each dictionary contains rule name, total strings, referenced strings,
    condition ratio, and file path.
    """
    try:
        # Initialize yaramod
        ymod = yaramod.Yaramod()
        # Parse the YARA file
        yara_file_parsed = ymod.parse_file(file_path)
        rules = yara_file_parsed.rules

        # List to store rule metrics
        rule_data = []

        print(f"\nProcessing file: {file_path}")
        print(f"Number of rules: {len(rules)}")

        for rule in rules:
            # print(f"\nRule name: {rule.name}")
            # Get all strings in the rule
            total_strings = len(rule.strings)
            # print(f"Total strings: {total_strings}")

            # Extract strings referenced in the condition
            # Assume all strings are referenced for ParenthesesExpression
            # (Extend for complex conditions if needed)
            referenced_strings = set()
            for string in rule.strings:
                referenced_strings.add(string.identifier)

            num_referenced = len(referenced_strings)
            condition_ratio = num_referenced / total_strings if total_strings > 0 else 0

            # print(f"Referenced strings in condition: {num_referenced}")
            # print(f"Condition ratio: {condition_ratio:.2f}")

            # Store rule metrics
            rule_data.append({
                "rule_name": rule.name,
                "total_strings": total_strings,
                "referenced_strings": num_referenced,
                "condition_ratio": condition_ratio,
                "file_path": file_path
            })

        return rule_data

    except Exception as e:
        print(f"Error processing file {file_path}: {str(e)}")
        return []

def main(file_path):
    # Specify the path to your YARA file

    # Collect data from the YARA file
    all_rule_data = analyze_yara_file(file_path)

    # Optional: Process multiple files in a directory
    """
    yara_dir = "/path/to/yara/rules/directory"
    all_rule_data = []
    for filename in os.listdir(yara_dir):
        if filename.endswith(".yar"):
            file_path = os.path.join(yara_dir, filename)
            rule_data = analyze_yara_file(file_path)
            all_rule_data.extend(rule_data)
    """

    # Create DataFrame
    if all_rule_data:
        df = pd.DataFrame(all_rule_data)
        # print("\nDataFrame created with the following columns:")
        # print(df.columns.tolist())
        # print("\nFirst few rows of the DataFrame:")
        # print(df.head())

        # Save DataFrame to CSV and pickle
        #output_csv = "yara_rule_metrics.csv"
        #output_pickle = "yara_rule_metrics.pkl"
        #df.to_csv(output_csv, index=False)
        # df.to_pickle(output_pickle)
    else:
        print("\nNo data to create DataFrame.")

    return df if all_rule_data else None

```

```python

def compute_metrics(df):
    """
    Compute Average Condition Ratio, Average Strings per Condition, and Conditions Count
    from the YARA rules DataFrame.
    """
    # Calculate metrics
    avg_condition_ratio = df["condition_ratio"].mean()
    avg_strings_per_condition = df["referenced_strings"].mean()
    conditions_count = len(df)  # One condition per rule

    # Create a dictionary with the results
    metrics = {
        "Average Condition Ratio": avg_condition_ratio,
        "Average Strings per Condition": avg_strings_per_condition,
        "Conditions Count": conditions_count
    }

    # Print results
    print("\nAggregated Metrics Across All Data:")
    print(f"Average Condition Ratio: {avg_condition_ratio:.2f}")
    print(f"Average Strings per Condition: {avg_strings_per_condition:.2f}")
    print(f"Conditions Count: {conditions_count}")

    return metrics
```
