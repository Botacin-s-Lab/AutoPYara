```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from itertools import islice
```

```python

def read_json_file(json_file, chunk_size=100000):
    """Reads the results_map.json file and returns a DataFrame, processing in chunks."""
    try:
        with open(json_file, 'r') as f:
            data = json.load(f)
    except Exception as e:
        raise ValueError(f"Error reading JSON file {json_file}: {e}")
    
    records = []
    for key, results in islice(data.items(), None):
        cluster, rule_id = key.split(",rule_")
        for result in results:
            result['cluster_name'] = cluster
            result['rule_id'] = rule_id
            records.append(result)
            if len(records) >= chunk_size:
                yield pd.DataFrame(records)
                records = []
    if records:
        yield pd.DataFrame(records)


def loadData(json_file):
    print("Reading JSON file...")
    df_chunks = read_json_file(json_file)
    df = pd.concat([chunk for chunk in df_chunks], ignore_index=True)
    return df
```
