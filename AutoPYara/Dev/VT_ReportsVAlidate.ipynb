```python

# Create an empty DataFrame to store the results
recovered_report_df = pd.DataFrame(columns=['File_Name', 'Full_Path', 'Matched_VT_Report'])

# List to collect rows for appending
new_rows = []

# Use tqdm for the progress bar in the loop
for _, row in tqdm(df_big_filtered.iterrows(), total=len(df_big_filtered), desc="Processing files"):
    file_name = row['File_Name']
    full_path = row['Full_Path']
    
    # Check for a similar report in missing_paths (partial match)
    for missing_path in missing_paths:
        if file_name.lower() in missing_path.lower():  # Similarity check
            # Add the matched row as a dictionary to the list
            new_rows.append({
                'File_Name': file_name,
                'Full_Path': full_path,
                'Matched_VT_Report': missing_path
            })

# If we have any new rows, create a new DataFrame
if new_rows:
    recovered_report_df = pd.concat([recovered_report_df, pd.DataFrame(new_rows)], ignore_index=True)

# Show the results
print(f"Found {len(recovered_report_df)} matches.")
```
